## Dataframes

We can create dataframes from multiple python datatypes.  Let's start with data stored in a dictionary where the key is a column name and the values are list of data. 

```python
data = {'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada'], 
        'year': [2000, 2001, 2002, 2001, 2002],
        'pop': [1.5, 1.7, 3.6, 2.4, 2.9]}
```

```python
test_data_frame = DataFrame(data)
test_data_frame
```

We can change the order of the columns by passing in the order.

```python
# Change column sequence
test_data_frame = DataFrame(data, columns=['year', 'state', 'pop'])
test_data_frame
```

If we pass a column name that does not exist in our data, the column will become NaN.  

```python
test_data_frame = DataFrame(data, columns=['year', 'state', 'pop','debt'])
test_data_frame
```

We can also specify index, or row names.

```python
# Change row names just like in Series
test_data_frame = DataFrame(data, columns=['year', 'state', 'pop'], 
                            index=['row one', 'row two', 'row three', 'row four', 'row five'])
test_data_frame
```

I can use a list of column names to query a dataframe for specific columns

```python
test_data_frame[['year', 'state']]
```

I can assign values to columns in multiple ways

```python
test_data_frame['debt'] = 25
test_data_frame
```

```python
# Assigning a column a vector value
test_data_frame['debt'] = np.arange(1,6,1)
test_data_frame
```

```python
# pass a series
new_data = Series([1.2, 2.4, 5.7],index=['row two', 'row three', 'row five'])
test_data_frame['debt'] = new_data
test_data_frame
```

```python
# Row shuffle
test_data_frame.reindex(['row five', 'row four', 'row three', 'row two', 'row one'])
```

We can do mathematical operations like summing rows and columns 

```python
test_data_frame.sum() # sum columns
```

```python
test_data_frame.sum(axis=1,numeric_only=True) # sum rows, skips over nonnumeric cells
```

We can also join two dataframes together. 

```python

data_frame_1 = DataFrame({'lkey': ['b', 'b', 'a', 'c', 'a', 'a', 'b'], 
                          'data1': range(7)})
data_frame_1
```

```python
data_frame_2 = DataFrame({'rkey': ['a', 'b', 'd'],
                          'data2': range(3)})
data_frame_2
```

```python
pd.merge(data_frame_1, data_frame_2,left_on='lkey', right_on='rkey')
```

## 3D-HST dataset
3D-HST is a near-infrared spectroscopic survey with the Hubble Space Telescope designed to study the physical processes that shape galaxies in the distant Universe.

3D-HST is surveying ~600 square arcminutes of well-studied extragalactic survey fields (AEGIS, COSMOS, GOODS-S, UKIDSS-UDS) from 248 orbits of Hubble Space Telescope within cycles 18 and 19. 

3D-HST now provides the critical third dimension - redshift - for ~10,000 galaxies at z>1

This is the epoch when ~60% of the star formation in the Universe took place, the number density of quasars peaked, the first galaxies stopped forming stars and the structural regularity that we see in galaxies today emerged. 


More info at: https://archive.stsci.edu/prepds/3d-hst/

```python
tab = tab[tab.lmass > 9].copy()
```

```python
target = 'z_spec'
features = [col for col in tab.columns if (col != target)]
```

```python
fig = plt.figure(0, [20,18])

for k, feat in enumerate(features):
    ax = fig.add_subplot(7, 6, k+1)
    ax.hist(tab[feat], bins=50, log=True, color='grey')
    ax.set_title(feat)

ax = fig.add_subplot(7, 6, len(features)+1)
ax.hist(tab[target], bins=50, color='red')
ax.set_title(r'$z_{\rm spec}$')

plt.tight_layout()
plt.show()
```

The ‘Av’, ‘lmass’ and ‘z_peak’ values were all computed via FAST photometric fit, and so we will exclude them as well. In addition, we will exclude the categorical flag variables (‘flags’, ‘f140w_flag’, ‘star_flag’, ‘use_phot’, ‘near_star’).

```python
features = [col for col in features if (col != 'Av') and (col != 'lmass') and (col != 'z_peak') 
            and (col != 'flags') and (col != 'f140w_flag') and (col != 'star_flag') 
            and (col != 'use_phot') and (col != 'near_star')]
```

Next, we will remove sources which have no constraints for our target variable:

```python
tab = tab[(tab[target] > 0)] 
```

Next, we will encode the ‘field” variable,

```python
label_encoder = LabelEncoder()
tab['field'] = label_encoder.fit_transform(tab['field'])
```

Finally we will impute missing values of photometric errors (set to -99. in the table) by assigning them the median of the distribution:

```python
errors = [col for col in features if (col[:1] == 'e') and (col[-1:] == 'W')]

for error in errors:
    tab[error] = np.where(tab[error] < -90, tab[error].median(), tab[error])
```

```python
fig = plt.figure(0, [20, 18])

for k, feat in enumerate(features):
    ax = fig.add_subplot(7, 6, k+1)
    ax.hist(tab[feat], bins=50, log=True, color='navy')
    ax.set_title(feat)

ax = fig.add_subplot(7, 6, len(features)+1)
ax.hist(tab[target], bins=50, color='red')
ax.set_title(r'$z_{\rm spec}$')

plt.tight_layout()
plt.show()
```

## Supernova dataset
The simulations solve an explosion inside a compression of a monatomic ideal gas, which follows the equation of state with the specific heat ratio γ=5/3. The gas in these simulations mocks the interstellar medium in the Milky Way Galaxy. 
At the beginning of the simulations, the thermal energy of a supernova is dumped at the center of the simulation box. The hot (\(\sim 10^7\) K) gas is immediately accelerated and makes the blastwave. Because velocities of the hot gas become supersonic, much fine resolution and small timestep are required to resolve the dynamics. 

The physical quantities are also distributed in seven orders of magnitude, which requires a large number of simulation steps.


###Dataset structure
- 59 time-steps of 64×64×64 cubes.

Fields available in the data: Pressure (scalar field), density (scalar field), temperature(scalar field), velocity (tensor field).

Number of trajectories: 740


```python
dataset_path = '/scratch1/10386/lsmith9003/share/tutorialData/supernova/supernova_explosion_Msun_0.1_dim64_file_10.hdf5'
```

```python
f = h5py.File(dataset_path,'r')


# top level keys
print(f.keys())
```

```python
# Reminder: 't0_fields', 't1_fields', 't2_fields' are respectively scalar fields, vector fields and tensor fields
# print the different fields available in the dataset
print("t0_fields:", f["t0_fields"].keys())
print("t1_fields:", f["t1_fields"].keys())
print("t2_fields:", f["t2_fields"].keys())
```

```python
# The data is of shape (n_trajectories, n_timesteps, x, y, z)
# Get the first t0_field and save it as a numpy array

traj = 3  # select the trajectory as the data is quite big

temperature = f["t0_fields"]["temperature"][traj, :]# HDF5 datasets can be sliced like a numpy array

print("shape of the selected t0_field: ", temperature.shape)
```

```python
from matplotlib.colors import LogNorm

x_slice = 32  # select the middle slice
traj_toplot = temperature[:, x_slice, :, :]

# field is now of shape (n_timesteps, y, z).
# Let's do a subplot to plot it at t= 0, t= T/3, t= 2T/3 and t= T:
fig, axs = plt.subplots(1, 4, figsize=(20, 5))
T = traj_toplot.shape[0]

# fix colorbar for all subplots:
normalize_plots = True
cmap = "magma"

if normalize_plots:
    vmin = np.min(traj_toplot)
    vmax = np.max(traj_toplot)
    norm = LogNorm(vmin=vmin, vmax=vmax)
    for i, t in enumerate([0, T // 3, (2 * T) // 3, T - 1]):
        axs[i].imshow(traj_toplot[t], cmap=cmap, norm=norm)
        axs[i].set_xticks([])
        axs[i].set_yticks([])
        axs[i].set_title(f"t={t}")
else:
    for i, t in enumerate([0, T // 3, (2 * T) // 3, T - 1]):
        axs[i].imshow(np.log(traj_toplot[t]), cmap=cmap)
        axs[i].set_xticks([])
        axs[i].set_yticks([])
        axs[i].set_title(f"t={t}")
plt.tight_layout()
```
