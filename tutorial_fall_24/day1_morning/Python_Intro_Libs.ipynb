# Python Data Types 

## Basic Variable Types 

The basic data types in python are 

- integers
- floats 
- strings
- booleans
- complex

Python data types are dynamically inferred. Python distinguishes between integers and floats by whether the number contains a decimal place.

```python
# Integer
apple_count = 10
# Float
apple_weight = 10.
# String
apple_name = "fuji"
# Boolean
is_an_apple = True

print(type(apple_count),type(apple_weight),type(apple_name),type(is_an_apple))

print("Is " + apple_name + " an apple? " + str(is_an_apple))
```

```python
# Two ways to create comlex numbers

# The first way - using comlex function 
complex_number_1 = complex(4, 9)
print(complex_number_1)
print(type(complex_number_1))

# The second way - using typical complex notation
complex_number_2 = 3 + 5j
print(complex_number_2)
print(type(complex_number_2))
```

## Collections: List, Tuple, Dictionary

Python and most modern languages follow the convention that arrays and lists of things start with an index of 0 and go to the last element being the array lenght minus one.  This is how computers think.  It does take some time getting used to.

### List

A list is a collection of objects (int, floats, string, any other python object):
- ordered - items stay in the same indexed position
- changeable - can be changed after it is defined

List are denoted with square brackets, i.e. [ and ]

```python
numbers = [ "one", 2, 3. ]
print(numbers)
```

List elements are indexed starting at zero.  You can access elements as follows:

```python
numbers[2]
```

Lists are changeable, so we can do things like append a new value to the end of a list.  Note each item has its own type

```python
numbers.append("four")
numbers[2] = 5

print(numbers)
print(type(numbers[0]),type(numbers[1]),type(numbers[2]),type(numbers[3]))
```

We can also compute the length of a list:

```python
print(len(numbers))
```

### Tuple

Tuples are similar to list, but they are immutable.  That is, tuples are  
- ordered  
- unchangable

Typles are denoted with round brackets, i.e. ( and )

```python
# Tuple
my_constants = (3.14, "taxes", 42)
```

```python
# Tuple : access element
my_constants[1]
```

```python
# Get length of tuple
len(my_constants)
```

```python
# Tuple : no append method i.e immutable so you cannot add to the tuple
my_constants.append(4)
```

```python
# Nor can you change a value after it is defined

my_constants[1]="e"
```

### Dictionary

Dictionaries store data as key:value pairs.  They are commonly used as a simple way to structure data.  You can access elements in a dictionary via the key.  They are initilized with the currly brackets { and }, with the key and value separated by a colon : and each element in the dictionary separated by a comma

```python
# Dictionary {Key:Value, Key:Value....}
shopping_list = {"carrots":1.5, "bag of rice":2, "coffee":"espresso bag"}
```

```python
# Dictionary : access element by key
print(shopping_list["coffee"])
```

```python
# Add new element
shopping_list["sugar"]="1 lb"
print(shopping_list)

# Get length of dictionary
print(len(shopping_list))

```

```python
# Values can be other structures 

shopping_list["mixed_nuts"]={"weight": 8, "unit": "oz"}
# Get length of dictionary
print(len(shopping_list))
print(shopping_list)
```

```python
# thats hard to read so lets use pretty print which is called pprint in the module pprint
# note this is horrible abbrivations and dupications but its not going to change

from pprint import pprint
pprint(shopping_list)
```

```python
# You can also go through the keys to see if an element is in there

print(shopping_list.keys())
print("flour" in shopping_list.keys())
```

## if/else statements

In the demo below, I determine if there are no elements in the list or if there are even or odd number of elements.

```python
if len(numbers)==0:  # boolean operater == asks if elements are the same 
    print("empty list")
elif len(numbers)%2==0:
    print("Even number of elements")
    print("!")
else:
    print("Odd number of elements") 
```

## For loop : Print even index elements of the list

In the demo below I print the even index elements of the list

```python
for element in numbers:
    if numbers.index(element)%2==0:
        print(element)
    else:
        # pass does nothing
        pass
    print('test')
```

## While loop : Search for things in list

```python
# Example 1
i = 0
while i < 4:
    print(i)
    i+=1 
```

```python
# Example 2
i = 0
while i < 5:
    i+=1
    if i % 2 == 0:
        continue
    elif i == 3:
        break
    print(i)
```

## Defining Functions

Below we define a function to search a list for an element and return its index if found and -1 if not found.  Returning a negative number from a list search is commonly used to say it was not found

```python
def print_name(name):
    print('My name is {}'.format(name))
    
print_name('bob')
```

```python
# so we have a list of our family

family_members = ["Mom", "Dad", "Kenny", "Julia"]
```

```python
def search_list_for_element(element, search_list):
    index=0
    while index < len(search_list):
        if search_list[index] == element:
            break
        else:
            index+=1
            continue
    #If index is less than length, element is found
    if index < len(search_list):
        return index
    else:
        return -1
```

```python
print(family_members)
```

# Matplotlib

Matplotlib has two styles interfaces when using the library: 

- explicit "axes" interface
   * gives user full control over the figure 
   * harder to use, but gives user more control
- implicit "pyplot" interface
   * matplotlib infers what user wants
   * easier to use, but give user less control
   
Read the [docs](https://matplotlib.org/devdocs/users/explain/api_interfaces.html) for more information
   
## pyplot interface

```python
# For plots inline in notebook
%matplotlib inline 

# Import matplotlib
import matplotlib.pyplot as plt

# we will also need numpy
import numpy as np
```

In the demo below we create several plots with the same data.  Note that in the various demos we just change a few parameters to change things like the color of the graph, scatter versus plot, etc.

```python
# Setup X and Y values
x = np.linspace(0,25, 20)
y = np.sqrt(x)

# Create plot
plt.xlabel('This is X label')
plt.ylabel('This is Y label')
print(x)
print(y)
plt.plot(x,y);
```

```python
# Blue circles
plt.xlabel('This is X label')
plt.ylabel('This is Y label')
plt.plot(x,y,'o');
```

```python
# Red circles
plt.xlabel('This is X label')
plt.ylabel('This is Y label')
plt.plot(x,y,'ro');
```

```python
# Green Lines
plt.xlabel('This is X label')
plt.ylabel('This is Y label')
plt.plot(x,y, 'g-');
```

```python
# Dotted Lines
plt.xlabel('This is X label')
plt.ylabel('This is Y label')
plt.plot(x,y,'--');
```

```python
# Bar Plot (histogram)
plt.xlabel('This is X label')
plt.ylabel('This is Y label')
plt.bar(x,y);
```

## Axes interface

In the demo below we will create two plots in one figure using the axes interface.  The two plots will show a sine and cosine wave

```python
#get data for plot demo
x=np.arange(0,1,0.025)
y1=10*np.cos(2*np.pi*5*x)
y2=10*np.sin(2*np.pi*5*x)
```

```python
figure, axes = plt.subplots(2,1)
```

Above we instantiated a fig and axs object used to control a figure. The diagram below highlights what these two objects control:

![](img/axes_matplotlib.png)

`.tight_layout()` is an example of a fig method.

```python
figure2, axes2 = plt.subplots(2,2)
figure2.tight_layout()
```

```python
figure, axes = plt.subplots(2,1)
axes[0].plot(x,y1,'r-')
axes[1].plot(x,y2,'k--')
axes[1].set_xlabel('This is X label')
axes[0].set_ylabel('This is Y label 1')
axes[1].set_ylabel('This is Y label 2')
figure.tight_layout();
```
