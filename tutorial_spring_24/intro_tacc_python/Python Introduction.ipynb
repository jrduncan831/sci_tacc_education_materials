## Collections: List, Tuple, Dictionary

Python and most modern languages follow the convention that arrays and lists of things start with an index of 0 and go to the last element being the array lenght minus one.  This is how computers think.  It does take some getting used to.

### List

A list is a collection of objects (int, floats, string, any other python object):
- ordered - items stay in the same indexed position
- changeable - can be changed after it is defined

List are denoted with square barackets e.g. [ and ]

```python
numbers = [ "one", 2, 3. ]
print(numbers)
```

List elements are indexed starting at zero.  You can access elements as follows:

```python
numbers[2]
```

Lists are changeable, so we can do things like append a new value to the end of a list.  Note each item has its own type

```python
numbers.append("four")
numbers[2] = 5

print(numbers)
print(type(numbers[0]),type(numbers[1]),type(numbers[2]),type(numbers[3]))
```

We can also compute the length of a list:

```python
print(len(numbers))
```

### Tuple

Tuples are similar to list, but the are immutable.  That is, tuples are  
- ordered  
- unchangable

Typles are defined using () 

```python
# Tuple
my_constants = (3.14, "taxes", 42)
```

```python
# Tuple : access element
my_constants[1]
```

```python
# Get length of tuple
len(my_constants)
```

```python
# Tuple : no append method i.e immutable so you cannot add to the tuple
my_constants.append(4)
```

```python
# Nor can you change a value after it is defined

my_constants[1]="e"
```

### Dictionary

Dictionaries store data as key:value pairs.  They are commonly used as a simple way to structure data.  You can access elements in a dictionary via the key.  They are initilized with the currly brackets { and }, with the key and value separated by a colon : and each element in the dictionary separated by a comma

```python
# Dictionary {Key:Value, Key:Value....}
shopping_list = {"carrots":1.5, "bag of rice":2, "coffee":"espresso bag"}
```

```python
# Dictionary : access element by key
print(shopping_list["coffee"])
```

```python
# Add new element
shopping_list["sugar"]="1 lb"
print(shopping_list)

# Get length of dictionary
print(len(shopping_list))

```

```python
# Values can be other structures 

shopping_list["mixed_nuts"]={"weight": 8, "unit": "oz"}
# Get length of dictionary
print(len(shopping_list))
print(shopping_list)
```

```python
# thats hard to read so lets use pretty print which is called pprint in the module pprint
# note this is horrible abbrivations and dupications but its not going to change

from pprint import pprint
pprint(shopping_list)
```

```python
# You can also go through the keys to see if an element is in there

print(shopping_list.keys())
print("flour" in shopping_list.keys())
```

## if/else statements

In the demo below, I determine if there are no elements in the list or if there are even or odd number of elements.

```python
if len(numbers)==0:  # boolean operater == asks if elements are the same 
    print("empty list")
elif len(numbers)%2==0:
    print("Even number of elements")
    print("!")
else:
    print("Odd number of elements") 
```

## While loop : Search for things in list

```python
# so we have a list of our family

family_members = ["Mom","Dad","Kenny"]

# we can append to this list
family_members.append("Julia")
print(family_members)
```

```python
# Now we want to see if Kenny is in there
# one way to do this is to loop over the array until
# we find the element.  


person = "Kennith"
if person in family_members:
    print("Found " + person)
else:
    print(person + " was not found")
    
person2 = "Kenny"
if person2 in family_members:
    print("Found " + person2)
else:
    print(person2 + " was not found")
```

## Defining Functions

Below we define a function to search a list for an element and return its index if found and -1 if not found.  Returning a negative number from a list search is commonly used to say it was not found

```python
def print_name(name):
    print('My name is {}'.format(name))
    
print_name('bob')
```

```python
def search_list_for_element(element, search_list):
    index=0
    while index < len(search_list):
        if search_list[index] == element:
            break
        else:
            index+=1
            continue
    #If index is less than length, element is found
    if index < len(search_list):
        return index
    else:
        return -1
```
